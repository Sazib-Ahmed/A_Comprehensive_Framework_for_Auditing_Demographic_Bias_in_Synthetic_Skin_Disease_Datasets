```python
# =================================================================================================
# Multimodal DINOv3 with Concatenation Fusion (Baseline for Ablation Study)
# - Uses 7-Fold CV ≈ 70% Train / 15% Val / 15% Test Overall
# - Serves as a direct comparison against more complex fusion methods like DSAF.
# - MODIFIED: Integrated a comprehensive visualization and metrics pipeline for research-grade analysis,
#   matching the reporting structure of the DSAF model for fair comparison.
# =================================================================================================

import os, time, json, warnings
from copy import deepcopy
from collections import defaultdict
from itertools import cycle
import numpy as np
import pandas as pd

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, Dataset
from torchvision import transforms
from PIL import Image

from sklearn.model_selection import StratifiedKFold, StratifiedShuffleSplit
from sklearn.metrics import (
    classification_report, accuracy_score, cohen_kappa_score,
    matthews_corrcoef, roc_auc_score, confusion_matrix, roc_curve, auc
)
from sklearn.preprocessing import label_binarize

# Conditional import for transformers
try:
    from transformers import AutoModel
    HAVE_TRANSFORMERS = True
except ImportError:
    HAVE_TRANSFORMERS = False

from torch.amp import GradScaler, autocast

warnings.filterwarnings('ignore')
torch.backends.cudnn.benchmark = True

# Optional FLOPs (best effort)
try:
    from thop import profile
    HAVE_THOP = True
except Exception:
    HAVE_THOP = False

# Optional plotting libs
try:
    import matplotlib.pyplot as plt
    import seaborn as sns
    HAVE_PLOTTING = True
except ImportError:
    HAVE_PLOTTING = False


# ---------------------------------------------
# 1) CONFIG — UPDATE THESE TWO PATHS
# ---------------------------------------------
class Config:
    DATA_DIR = r"D:\AIUB\Neural\Synthetic_Skin_Disease_Dataset"     # <-- update to your root image dir
    ORIGINAL_METADATA_CSV = r"D:\AIUB\Neural\complete_metadata_m.csv" # <-- update to your CSV path

    RESULTS_DIR = "metadata_Multimodal_DINOv3_Concat_KFold" # <-- Directory for this baseline
    MODEL_NAME = "facebook/dinov3-vitb16-pretrain-lvd1689m"
    IMG_SIZE = 224
    BATCH_SIZE = 128
    LEARNING_RATE = 1e-4
    WEIGHT_DECAY = 2e-5
    EPOCHS = 200
    NUM_CLASSES = 5
    DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
    USE_AMP = True
    NUM_WORKERS = 0
    PATIENCE = 20
    WARMUP_EPOCHS = 10
    LABEL_SMOOTHING = 0.1
    PARTIALLY_UNFREEZE = False # set True to unfreeze last block of DINOv3

    # K-Fold settings for ~70/15/15 overall
    N_SPLITS = 7
    VAL_WITHIN_REMAINDER = 0.15 / (1.0 - 1.0/7.0) # ≈ 0.175

    # Metadata settings
    META_DROPOUT = 0.3
    META_HIDDEN_DIM = 128
    META_OUT_DIM = 256

    # Fairness: minimum subgroup size for stable metrics
    MIN_SUPPORT = 15

config = Config()
os.makedirs(config.RESULTS_DIR, exist_ok=True)
print("Device:", config.DEVICE)
print(f"Model: DINOv3 with Concatenation Fusion (Baseline)")
print(f"K-Fold: N_SPLITS={config.N_SPLITS} (~Test {100/config.N_SPLITS:.1f}%), Val_in_remainder={config.VAL_WITHIN_REMAINDER:.3f}")
if not HAVE_PLOTTING:
    print("Warning: matplotlib and/or seaborn not found. Plots will be skipped.")


# ---------------------------------------------
# 2) CSV STANDARDIZATION
# ---------------------------------------------
def standardize_metadata_columns(df: pd.DataFrame) -> pd.DataFrame:
    """Standardizes metadata column names and creates necessary fields."""
    df = df.copy()
    if 'filepath' in df.columns:
        df['filepath'] = df['filepath'].astype(str).str.replace('\\', '/', regex=False)
    if 'filename' not in df.columns:
        if 'filepath' in df.columns:
            df['filename'] = df['filepath'].apply(lambda p: os.path.basename(str(p)))
        else:
            raise ValueError("CSV must include 'filepath' or 'filename'.")
    if 'label' not in df.columns and 'filepath' in df.columns:
        df['label'] = df['filepath'].str.split('/').str[1]
    df['label'] = df['label'].astype(str)
    if 'age_group' not in df.columns:
        if 'age' in df.columns:
            age_num = pd.to_numeric(df['age'], errors='coerce')
            bins    = [-1, 17, 29, 44, 64, 200]
            names   = ['child', 'young_adult', 'adult', 'middle_aged', 'senior']
            df['age_group'] = pd.cut(age_num, bins=bins, labels=names).astype(str).fillna('unknown')
        else:
            df['age_group'] = 'unknown'
    if 'gender' not in df.columns:
        if 'dominant_gender' in df.columns:
            g = df['dominant_gender'].astype(str).str.strip().str.lower()
            df['gender'] = g.map({'man':'male','male':'male','woman':'female','female':'female'}).fillna('unknown')
        else:
            df['gender'] = 'unknown'
    if 'skin_tone' not in df.columns:
        if 'dominant_race' in df.columns:
            df['skin_tone'] = df['dominant_race'].astype(str).str.strip().str.lower()
        else:
            df['skin_tone'] = 'unknown'
    required = ['filename', 'label', 'age_group', 'gender', 'skin_tone']
    if any(c not in df.columns for c in required):
        raise ValueError(f"Missing required columns after standardization.")
    return df


# ---------------------------------------------
# 3) DATASET & VOCAB
# ---------------------------------------------
class GPUAugOnlyNorm(nn.Module):
    """Module for GPU-side normalization."""
    def __init__(self):
        super().__init__()
        self.normalize = transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    @torch.no_grad()
    def forward(self, x): return self.normalize(x)

class MetaVocab:
    """Creates and manages vocabularies for metadata fields."""
    def __init__(self, df_train, cols=('age_group','gender','skin_tone')):
        self.cols = cols
        self.unk_token = "<UNK>"
        self.vocabs = {}
        for c in cols:
            uniq = sorted([self.unk_token] + sorted(df_train[c].dropna().astype(str).unique().tolist()))
            self.vocabs[c] = {v:i for i,v in enumerate(uniq)}
        self.sizes = {c: len(self.vocabs[c]) for c in cols}
    def encode_row(self, row):
        out = {}
        for c in self.cols:
            v = str(row[c]) if pd.notnull(row[c]) else self.unk_token
            out[c] = self.vocabs[c].get(v, self.vocabs[c][self.unk_token])
        return out

class MultimodalDermDataset(Dataset):
    """Dataset class for images and metadata, with optional preloading."""
    def __init__(self, config, df, class_to_idx, meta_vocab, transform=None):
        self.config = config
        self.df = df.reset_index(drop=True)
        self.class_to_idx = class_to_idx
        self.transform = transform
        self.meta_vocab = meta_vocab
        self.image_paths, self.labels, self.meta = [], [], []
        use_filepath = 'filepath' in self.df.columns
        for _, row in self.df.iterrows():
            if use_filepath and pd.notnull(row.get('filepath', None)):
                raw = str(row['filepath'])
                p = raw if os.path.isabs(raw) else os.path.join(config.DATA_DIR, raw)
            else:
                p = os.path.join(config.DATA_DIR, str(row['split']), str(row['label']), str(row['filename']))
            self.image_paths.append(p)
            self.labels.append(self.class_to_idx[str(row['label'])])
            self.meta.append(self.meta_vocab.encode_row(row))
        self.images = []
        for p in self.image_paths:
            with Image.open(p).convert("RGB") as im:
                self.images.append(im.copy())

    def __len__(self): return len(self.labels)
    def __getitem__(self, idx):
        image = self.images[idx]
        if self.transform: image = self.transform(image)
        return image, self.labels[idx], self.meta[idx]


# ---------------------------------------------
# 4) MODEL (DINOv3 + Metadata Concatenation)
# ---------------------------------------------
class MetadataEncoder(nn.Module):
    """Encodes categorical metadata into a dense feature vector."""
    def __init__(self, meta_vocab, emb_dims=None, hidden=128, out_dim=256, dropout=0.1):
        super().__init__()
        sizes = meta_vocab.sizes
        if emb_dims is None:
            emb_dims = {
                'age_group': min(16, max(4, sizes['age_group'])),
                'gender':    min(8,  max(3, sizes['gender'])),
                'skin_tone': min(16, max(4, sizes['skin_tone'])),
            }
        self.emb_age = nn.Embedding(sizes['age_group'], emb_dims['age_group'])
        self.emb_gen = nn.Embedding(sizes['gender'],    emb_dims['gender'])
        self.emb_tone= nn.Embedding(sizes['skin_tone'], emb_dims['skin_tone'])
        concat_dim = sum(emb_dims.values())
        self.mlp = nn.Sequential(
            nn.Linear(concat_dim, hidden), nn.GELU(), nn.Dropout(dropout),
            nn.Linear(hidden, out_dim), nn.GELU()
        )
    def forward(self, meta_idx):
        a = self.emb_age(meta_idx['age_group'])
        g = self.emb_gen(meta_idx['gender'])
        s = self.emb_tone(meta_idx['skin_tone'])
        return self.mlp(torch.cat([a,g,s], dim=-1))

class DINOv3MultimodalConcat(nn.Module):
    """
    A multimodal model using a pretrained DINOv3 backbone.
    Features from the image backbone's [CLS] token are concatenated
    with features from the metadata encoder and passed to a final classifier.
    This serves as a strong but simple multimodal baseline.
    """
    def __init__(self, model_name, num_classes, meta_vocab, meta_hidden_dim, meta_out_dim, meta_dropout):
        super().__init__()
        if not HAVE_TRANSFORMERS:
            raise ImportError("Please install transformers: pip install transformers")

        # Image backbone (DINOv3)
        self.backbone = AutoModel.from_pretrained(model_name)

        # Freeze all backbone parameters initially
        for param in self.backbone.parameters():
            param.requires_grad = False

        # Unfreeze the final transformer block if configured
        if config.PARTIALLY_UNFREEZE and hasattr(self.backbone, "vit") and hasattr(self.backbone.vit, "encoder"):
            for param in self.backbone.vit.encoder.layer[-1].parameters():
                param.requires_grad = True

        image_feature_dim = self.backbone.config.hidden_size # e.g., 768 for ViT-B

        # Metadata encoder
        self.meta_encoder = MetadataEncoder(meta_vocab,
                                            hidden=meta_hidden_dim,
                                            out_dim=meta_out_dim,
                                            dropout=meta_dropout)

        # Fusion and Classifier Head
        self.classifier = nn.Sequential(
            nn.Linear(image_feature_dim + meta_out_dim, 512),
            nn.BatchNorm1d(512),
            nn.GELU(),
            nn.Dropout(0.5),
            nn.Linear(512, num_classes)
        )

    def forward(self, pixel_values, meta_idx):
        self.backbone.eval() if not config.PARTIALLY_UNFREEZE else None
        with torch.no_grad() if not config.PARTIALLY_UNFREEZE else torch.enable_grad():
            outputs = self.backbone(pixel_values=pixel_values)

        img_features = outputs.last_hidden_state[:, 0]
        meta_features = self.meta_encoder(meta_idx)
        combined_features = torch.cat((img_features, meta_features), dim=1)
        output = self.classifier(combined_features)
        return output


# ---------------------------------------------
# 5) TRAIN / EVAL HELPERS
# ---------------------------------------------
def _to_device_meta(batch_meta, device):
    if isinstance(batch_meta, dict):
        ag = torch.as_tensor(batch_meta['age_group'], dtype=torch.long, device=device)
        ge = torch.as_tensor(batch_meta['gender'],    dtype=torch.long, device=device)
        st = torch.as_tensor(batch_meta['skin_tone'], dtype=torch.long, device=device)
        return {'age_group': ag, 'gender': ge, 'skin_tone': st}
    else:
        ag = torch.tensor([m['age_group'] for m in batch_meta], dtype=torch.long, device=device)
        ge = torch.tensor([m['gender']    for m in batch_meta], dtype=torch.long, device=device)
        st = torch.tensor([m['skin_tone'] for m in batch_meta], dtype=torch.long, device=device)
        return {'age_group': ag, 'gender': ge, 'skin_tone': st}

def make_dataloaders_from_df(config, df_std):
    """Creates dataloaders for Train, Val, and Test splits from a DataFrame."""
    classes = sorted(df_std[df_std['split']=='Train']['label'].astype(str).unique())
    class_to_idx = {c:i for i,c in enumerate(classes)}
    config.NUM_CLASSES = len(classes)
    meta_vocab = MetaVocab(df_std[df_std['split']=='Train'])
    train_tfm = transforms.Compose([
        transforms.Resize((config.IMG_SIZE, config.IMG_SIZE)),
        transforms.RandomHorizontalFlip(),
        transforms.RandomApply([transforms.ColorJitter(0.4,0.4,0.2,0.1)], p=0.8),
        transforms.RandomRotation(15),
        transforms.ToTensor(),
    ])
    eval_tfm = transforms.Compose([
        transforms.Resize((config.IMG_SIZE, config.IMG_SIZE)),
        transforms.ToTensor(),
    ])
    dsets = {s: MultimodalDermDataset(config, df_std[df_std['split']==s], class_to_idx, meta_vocab, t)
             for s, t in [('Train', train_tfm), ('Val', eval_tfm), ('Test', eval_tfm)]}
    dls = {s: DataLoader(dsets[s], batch_size=config.BATCH_SIZE, shuffle=(s=='Train'),
                         num_workers=config.NUM_WORKERS, pin_memory=True) for s in dsets}
    return dls, {k: len(v) for k,v in dsets.items()}, classes, meta_vocab

def plot_and_save_history(hist, fold_id, results_dir):
    """Plots training and validation loss and accuracy and saves them to a file."""
    if not HAVE_PLOTTING: return

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 6))
    epochs = range(1, len(hist['train_loss']) + 1)
    ax1.plot(epochs, hist['train_loss'], 'bo-', label='Training loss')
    ax1.plot(epochs, hist['val_loss'], 'ro-', label='Validation loss')
    ax1.set_title(f'Training and Validation Loss (Fold {fold_id})'); ax1.set_xlabel('Epochs'); ax1.set_ylabel('Loss')
    ax1.legend(); ax1.grid(True)
    ax2.plot(epochs, hist['train_acc'], 'bo-', label='Training accuracy')
    ax2.plot(epochs, hist['val_acc'], 'ro-', label='Validation accuracy')
    ax2.set_title(f'Training and Validation Accuracy (Fold {fold_id})'); ax2.set_xlabel('Epochs'); ax2.set_ylabel('Accuracy')
    ax2.legend(); ax2.grid(True)
    plt.tight_layout()
    history_path = os.path.join(results_dir, f'training_history_fold_{fold_id}.png')
    plt.savefig(history_path, dpi=300); plt.show(); plt.close()
    print(f"Saved training history plot to: {history_path}")

def train_model(model, dls, sizes, num_epochs, fold_id):
    start = time.time()
    best = deepcopy(model.state_dict())
    best_val_loss = float('inf')
    patience = 0
    history = defaultdict(list)
    scaler = GradScaler(enabled=config.USE_AMP)
    gpu_norm = GPUAugOnlyNorm().to(config.DEVICE)
    optimizer = optim.AdamW(filter(lambda p: p.requires_grad, model.parameters()),
                            lr=config.LEARNING_RATE, weight_decay=config.WEIGHT_DECAY)
    def lr_lambda(ep):
        if ep < config.WARMUP_EPOCHS: return float(ep+1)/config.WARMUP_EPOCHS
        return max(0.0, 0.5 * (1.0 + np.cos(np.pi * (ep - config.WARMUP_EPOCHS) / (num_epochs - config.WARMUP_EPOCHS))))
    scheduler = optim.lr_scheduler.LambdaLR(optimizer, lr_lambda)
    criterion = nn.CrossEntropyLoss(label_smoothing=config.LABEL_SMOOTHING)
    best_model_path = os.path.join(config.RESULTS_DIR, f'best_model_fold_{fold_id}.pth')

    print(f"\nSTART TRAINING (DINOv3 + Concat) for {num_epochs} epochs...")
    for epoch in range(num_epochs):
        for phase in ['Train', 'Val']:
            model.train(phase == 'Train')
            running_loss, running_corrects = 0.0, 0
            for images, labels, metas in dls[phase]:
                images = gpu_norm(images.to(config.DEVICE))
                labels = labels.to(config.DEVICE, dtype=torch.long)
                meta_idx = _to_device_meta(metas, config.DEVICE)
                optimizer.zero_grad(set_to_none=True)
                with autocast(device_type=config.DEVICE.split(':')[0], enabled=config.USE_AMP):
                    logits = model(pixel_values=images, meta_idx=meta_idx)
                    loss = criterion(logits, labels)
                if phase == 'Train':
                    scaler.scale(loss).backward(); scaler.step(optimizer); scaler.update()
                running_loss += loss.item() * images.size(0)
                running_corrects += (logits.argmax(1) == labels).sum().item()

            epoch_loss = running_loss / sizes[phase]
            epoch_acc = running_corrects / sizes[phase]
            history[f'{phase.lower()}_loss'].append(epoch_loss)
            history[f'{phase.lower()}_acc'].append(epoch_acc)

            if phase == 'Val':
                if epoch_loss < best_val_loss:
                    best_val_loss = epoch_loss; best = deepcopy(model.state_dict()); patience = 0
                    torch.save(model.state_dict(), best_model_path)
                else:
                    patience += 1
        scheduler.step()
        print(f"Epoch {epoch+1}/{num_epochs} | Train Loss: {history['train_loss'][-1]:.4f} Acc: {history['train_acc'][-1]:.4f} | "
              f"Val Loss: {history['val_loss'][-1]:.4f} Acc: {history['val_acc'][-1]:.4f} | LR: {scheduler.get_last_lr()[0]:.2e}")
        if patience >= config.PATIENCE:
            print("Early stopping triggered."); break

    print(f"Training complete in {(time.time()-start)/60.0:.1f} minutes. Best Val Loss: {best_val_loss:.4f}")
    print(f"Best model for fold saved to: {best_model_path}")
    model.load_state_dict(best)
    plot_and_save_history(history, fold_id, config.RESULTS_DIR)
    return model, history

# ---------------------------------------------
# 6) EVALUATION & FAIRNESS METRICS
# ---------------------------------------------
def _print_section(title):
    print("\n" + "="*80); print(title.upper()); print("="*80 + "\n")

def _group_fairness_metrics(y_true, y_pred, groups, num_classes, min_support=None):
    uniq_groups = sorted(pd.Series(groups).astype(str).unique())
    per_group_acc, per_group_tprs, counts = {}, {}, {}
    y_true, y_pred = np.asarray(y_true), np.asarray(y_pred)
    for g in uniq_groups:
        mask = (np.asarray(groups) == g); counts[g] = int(mask.sum())
        yt, yp = y_true[mask], y_pred[mask]
        per_group_acc[g] = (yt == yp).mean() if len(yt) > 0 else np.nan
        tprs = [( ((yt == c) & (yp == c)).sum() / ((yt == c).sum() or 1) ) for c in range(num_classes)]
        per_group_tprs[g] = np.array(tprs, dtype=float)
    filt_keys = [g for g in uniq_groups if counts[g] >= min_support] if min_support is not None else uniq_groups
    wga = float(np.nanmin([per_group_acc.get(g, np.nan) for g in filt_keys])) if filt_keys else float('nan')
    max_gap, worst_pair = 0.0, (None, None)
    for i in range(len(filt_keys)):
        for j in range(i + 1, len(filt_keys)):
            gi, gj = filt_keys[i], filt_keys[j]
            a, b = per_group_tprs[gi], per_group_tprs[gj]
            mask = ~(np.isnan(a) | np.isnan(b))
            if mask.sum() > 0:
                gap = float(np.nanmean(np.abs(a[mask] - b[mask])))
                if gap > max_gap: max_gap, worst_pair = gap, (gi, gj)
    return {'wga': wga, 'eod': max_gap, 'worst_pair': worst_pair, 'keys_used': filt_keys}

def plot_and_save_roc_curves(y_true_bin, y_prob, class_names, fold_id, results_dir):
    if not HAVE_PLOTTING: return
    n_classes = len(class_names)
    fpr, tpr, roc_auc = dict(), dict(), dict()
    for i in range(n_classes):
        fpr[i], tpr[i], _ = roc_curve(y_true_bin[:, i], y_prob[:, i])
        roc_auc[i] = auc(fpr[i], tpr[i])
    all_fpr = np.unique(np.concatenate([fpr[i] for i in range(n_classes)]))
    mean_tpr = np.zeros_like(all_fpr)
    for i in range(n_classes): mean_tpr += np.interp(all_fpr, fpr[i], tpr[i])
    mean_tpr /= n_classes
    fpr["macro"], tpr["macro"], roc_auc["macro"] = all_fpr, mean_tpr, auc(all_fpr, mean_tpr)
    plt.figure(figsize=(12, 10))
    colors = cycle(['aqua', 'darkorange', 'cornflowerblue', 'green', 'red', 'purple', 'brown'])
    for i, color in zip(range(n_classes), colors):
        plt.plot(fpr[i], tpr[i], color=color, lw=2, label=f'ROC of {class_names[i]} (AUC = {roc_auc[i]:0.2f})')
    plt.plot(fpr["macro"], tpr["macro"], label=f'Macro-average ROC (AUC = {roc_auc["macro"]:0.2f})', color='navy', linestyle=':', linewidth=4)
    plt.plot([0, 1], [0, 1], 'k--', lw=2); plt.xlim([0.0, 1.0]); plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate'); plt.ylabel('True Positive Rate'); plt.title(f'Multi-class ROC - Fold {fold_id}')
    plt.legend(loc="lower right"); plt.grid(True)
    roc_path = os.path.join(results_dir, f'roc_curves_fold_{fold_id}.png')
    plt.savefig(roc_path, dpi=300); plt.show(); plt.close()
    print(f"Saved ROC curves plot to: {roc_path}")

def plot_and_save_scalar_metrics(metrics_dict, fold_id, results_dir):
    if not HAVE_PLOTTING: return
    names, values = list(metrics_dict.keys()), list(metrics_dict.values())
    plt.figure(figsize=(10, 6))
    bars = sns.barplot(x=names, y=values, palette="viridis")
    plt.title(f'Overall Scalar Metrics Summary - Fold {fold_id}'); plt.ylabel('Score'); plt.ylim(0, 1.05)
    for bar in bars.patches:
        plt.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 0.01, f'{bar.get_height():.4f}', ha='center', va='bottom', fontsize=10)
    scalar_path = os.path.join(results_dir, f'scalar_metrics_fold_{fold_id}.png')
    plt.savefig(scalar_path, dpi=300); plt.show(); plt.close()
    print(f"Saved scalar metrics plot to: {scalar_path}")

def evaluate_and_visualize(model, dl_test, class_names, df_fold, fold_id):
    model.eval()
    gpu_norm = GPUAugOnlyNorm().to(config.DEVICE)
    y_true, y_pred, y_prob = [], [], []
    with torch.no_grad():
        for images, labels, metas in dl_test:
            images = gpu_norm(images.to(config.DEVICE))
            meta_idx = _to_device_meta(metas, config.DEVICE)
            logits = model(pixel_values=images, meta_idx=meta_idx)
            y_true.extend(labels.numpy())
            y_pred.extend(logits.argmax(1).cpu().numpy())
            y_prob.extend(torch.softmax(logits, dim=1).cpu().numpy())
    y_true, y_pred, y_prob = np.array(y_true), np.array(y_pred), np.array(y_prob)

    _print_section(f"OVERALL PERFORMANCE (FOLD {fold_id})")
    print(classification_report(y_true, y_pred, target_names=class_names, zero_division=0))

    acc = accuracy_score(y_true, y_pred)
    kappa = cohen_kappa_score(y_true, y_pred)
    mcc = matthews_corrcoef(y_true, y_pred)
    try:
        y_true_bin = label_binarize(y_true, classes=np.arange(len(class_names)))
        roc_auc_macro = roc_auc_score(y_true_bin, y_prob, average='macro', multi_class='ovr')
    except Exception: y_true_bin, roc_auc_macro = None, float('nan')
    print(f"\n--- Overall Scalar Metrics ---\nAccuracy: {acc:.4f}\nMacro ROC AUC: {roc_auc_macro:.4f}\nCohen's Kappa: {kappa:.4f}\nMatthews CorrCoef: {mcc:.4f}\n")

    if HAVE_PLOTTING:
        cm = confusion_matrix(y_true, y_pred, labels=np.arange(len(class_names)))
        cm_norm = np.nan_to_num(cm.astype('float') / cm.sum(axis=1)[:, np.newaxis])
        plt.figure(figsize=(12, 10))
        sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='Blues', xticklabels=class_names, yticklabels=class_names, vmin=0, vmax=1)
        plt.title(f'Normalized Confusion Matrix (Fold {fold_id})'); plt.ylabel('True Label'); plt.xlabel('Predicted Label'); plt.tight_layout()
        cm_path = os.path.join(config.RESULTS_DIR, f'confusion_matrix_fold_{fold_id}.png')
        plt.savefig(cm_path, dpi=300); plt.show(); plt.close()
        print(f"Saved normalized confusion matrix to: {cm_path}")
        if y_true_bin is not None: plot_and_save_roc_curves(y_true_bin, y_prob, class_names, fold_id, config.RESULTS_DIR)
        plot_and_save_scalar_metrics({'Accuracy': acc, 'Macro AUC': roc_auc_macro, 'Kappa': kappa, 'MCC': mcc}, fold_id, config.RESULTS_DIR)

    df_test = df_fold[df_fold['split'] == 'Test'].reset_index(drop=True)
    df_test = df_test.iloc[:len(y_true)] # Align rows just in case
    age_disp = df_test['age_group'].map({'child': '0-17', 'young_adult': '18-59', 'adult': '18-59', 'middle_aged': '18-59', 'senior': '60+'}).fillna('unknown')

    # *** FIX IS APPLIED HERE ***
    # Define the full set of labels to expect in any report.
    full_labels = np.arange(len(class_names))

    for group_name, series in [('GENDER', df_test['gender']), ('AGE', age_disp), ('SKIN TONE', df_test['skin_tone'])]:
        _print_section(f"PERFORMANCE BY {group_name}")
        for val in sorted(series.unique()):
            if val == 'unknown': continue
            mask = (series == val).values
            if mask.sum() > 0:
                print(f"--- Subgroup: {val} ---\n")
                # Add the 'labels' parameter to the classification_report call
                print(classification_report(y_true[mask], y_pred[mask],
                                            target_names=class_names,
                                            labels=full_labels, # This ensures all 5 classes are in the report
                                            zero_division=0))

    groups_composite = (age_disp.astype(str) + "|" + df_test['gender'].astype(str) + "|" + df_test['skin_tone'].astype(str))
    fair_all = _group_fairness_metrics(y_true, y_pred, groups_composite, len(class_names))
    fair_filt = _group_fairness_metrics(y_true, y_pred, groups_composite, len(class_names), min_support=config.MIN_SUPPORT)
    print("\n" + "="*80 + f"\nFAIRNESS SUMMARY (ALL GROUPS) | WGA: {fair_all['wga']:.4f} EOD: {fair_all['eod']:.4f}\nWorst pair: {fair_all['worst_pair'][0]} vs {fair_all['worst_pair'][1]}\n" + "="*80)
    print("\n" + "="*80 + f"\nFAIRNESS SUMMARY (FILTERED, n>={config.MIN_SUPPORT}) | WGA: {fair_filt['wga']:.4f} EOD: {fair_filt['eod']:.4f}\nWorst pair: {fair_filt['worst_pair'][0]} vs {fair_filt['worst_pair'][1]}\nGroups: {fair_filt['keys_used']}\n" + "="*80)

    def acc_counts(s, yt, yp): return {str(g): ((yt[s==g] == yp[s==g]).sum(), (s==g).sum()) for g in s.unique()}
    pooled_info = {'gender': acc_counts(df_test['gender'], y_true, y_pred), 'age': acc_counts(age_disp, y_true, y_pred), 'skin_tone': acc_counts(df_test['skin_tone'], y_true, y_pred), 'composite': acc_counts(groups_composite, y_true, y_pred)}

    return {'accuracy': acc, 'roc_auc_macro': roc_auc_macro, 'kappa': kappa, 'mcc': mcc, 'wga_all': fair_all['wga'], 'eod_all': fair_all['eod'], 'wga_filtered': fair_filt['wga'], 'eod_filtered': fair_filt['eod'], 'pooled_groups': pooled_info}

# ---------------------------------------------
# 7) K-FOLD CROSS-VALIDATION ORCHESTRATION
# ---------------------------------------------
def build_kfold_splits(df_std, n_splits, val_within_remainder, seed=42):
    df, y = df_std.copy().reset_index(drop=True), df_std['label'].astype(str).values
    skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=seed)
    folds = []
    for i, (trainval_idx, test_idx) in enumerate(skf.split(np.arange(len(df)), y), 1):
        sss2 = StratifiedShuffleSplit(n_splits=1, test_size=val_within_remainder, random_state=seed+i)
        tr_rel, val_rel = next(sss2.split(trainval_idx, y[trainval_idx]))
        dfi = df.copy(); dfi['split'] = 'Ignore'
        dfi.loc[trainval_idx[tr_rel], 'split'], dfi.loc[trainval_idx[val_rel], 'split'], dfi.loc[test_idx, 'split'] = 'Train', 'Val', 'Test'
        folds.append(dfi[dfi['split'] != 'Ignore'])
    return folds

def run_one_fold(fold_df, fold_id, config):
    print("\n" + "#"*88 + f"\n### FOLD {fold_id} ###\n" + "#"*88)
    dls, sizes, class_names, meta_vocab = make_dataloaders_from_df(config, fold_df)
    model = DINOv3MultimodalConcat(model_name=config.MODEL_NAME, num_classes=len(class_names), meta_vocab=meta_vocab, meta_hidden_dim=config.META_HIDDEN_DIM, meta_out_dim=config.META_OUT_DIM, meta_dropout=config.META_DROPOUT).to(config.DEVICE)
    print(f"Trainable params: {sum(p.numel() for p in model.parameters() if p.requires_grad)/1e6:.2f}M")
    model, _ = train_model(model, dls, sizes, num_epochs=config.EPOCHS, fold_id=fold_id)
    return evaluate_and_visualize(model, dls['Test'], class_names, fold_df, fold_id)

def kfold_cross_val(df_std, config):
    folds = build_kfold_splits(df_std, config.N_SPLITS, config.VAL_WITHIN_REMAINDER)
    per_fold = [run_one_fold(fold, i + 1, config) for i, fold in enumerate(folds)]
    perf = pd.DataFrame(per_fold)

    def mean_std(col): return (perf[col].mean(), perf[col].std())
    summary = {k: mean_std(k) for k in ['accuracy', 'roc_auc_macro', 'kappa', 'mcc', 'wga_all', 'eod_all', 'wga_filtered', 'eod_filtered']}

    print("\n" + "="*88 + f"\nCross-validated over {config.N_SPLITS} folds\n" + "="*88)
    for k, (m, s) in summary.items(): print(f"{k.replace('_', ' ').title():<20}: {m:.3f} ± {s:.3f}")

    def merge_pooled(key):
        agg = defaultdict(lambda: [0, 0])
        for m in per_fold:
            for g, (c, t) in m['pooled_groups'][key].items(): agg[g][0] += c; agg[g][1] += t
        return sorted([(g, c/t, t) for g,(c,t) in agg.items() if t>0], key=lambda x: -x[2])

    def print_table(title, rows):
        print(f"\n--- {title} ---"); print(f"{'Group':35s} {'Acc':>8s} {'Count':>8s}")
        for g, acc, cnt in rows:
            if cnt >= config.MIN_SUPPORT: print(f"{g:35s} {acc:8.3f} {cnt:8d}")

    for key, name in [('gender', 'GENDER'), ('age', 'AGE'), ('skin_tone', 'SKIN TONE'), ('composite', 'COMPOSITE')]:
        print_table(f"POOLED OVERALL ACCURACY BY {name} (n>={config.MIN_SUPPORT})", merge_pooled(key))

    perf['fold'] = range(1, len(perf) + 1); perf.set_index('fold', inplace=True)
    perf.to_csv(os.path.join(config.RESULTS_DIR, "kfold_performance.csv"))
    pd.DataFrame({k: {'mean': v[0], 'std': v[1]} for k,v in summary.items()}).T.to_json(os.path.join(config.RESULTS_DIR, "kfold_summary.json"), indent=2)
    print(f"\nResults, models, and plots saved to '{config.RESULTS_DIR}'")


# ---------------------------------------------
# 8) MAIN EXECUTION
# ---------------------------------------------
if __name__ == '__main__':
    try:
        df_orig = pd.read_csv(config.ORIGINAL_METADATA_CSV)
        df_std = standardize_metadata_columns(df_orig)
        print(f"Loaded and standardized metadata: {len(df_std)} rows.")
        kfold_cross_val(df_std, config)
    except FileNotFoundError:
        print(f"ERROR: Metadata file not found at '{config.ORIGINAL_METADATA_CSV}'")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
        import traceback
        traceback.print_exc()
```

```python
# =================================================================================================
# Attention Map Visualization Script for Multiple Subgroup Comparisons
# - Generates separate plots for different demographic axes (age, skin tone, etc.)
# - Easily configurable via the `plotting_jobs` list in the main section.
# =================================================================================================

import os
import random
import warnings
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
from PIL import Image
from torchvision import transforms
import cv2

# Conditional import for Hugging Face transformers
try:
    from transformers import AutoModel, AutoConfig
    HAVE_TRANSFORMERS = True
except ImportError:
    HAVE_TRANSFORMERS = False

# Suppress warnings for cleaner output
warnings.filterwarnings('ignore')
torch.backends.cudnn.benchmark = True


# =================================================================================================
# 1. CONFIGURATION
# =================================================================================================
class Config:
    DATA_DIR = r"D:\AIUB\Neural\Synthetic_Skin_Disease_Dataset"
    ORIGINAL_METADATA_CSV = r"D:\AIUB\Neural\complete_metadata_m.csv"
    RESULTS_DIR = "new_results_Multimodal_DINO_Concat_KFold"
    LOAD_MODEL_PATH = os.path.join(RESULTS_DIR, "best_model_fold_5.pth")
    MODEL_NAME = "facebook/dinov3-vitb16-pretrain-lvd1689m"
    IMG_SIZE = 224
    DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
    N_SPLITS = 7

config = Config()
print(f"Using device: {config.DEVICE}")


# =================================================================================================
# 2. MODEL AND DATA HELPERS (from previous versions)
# =================================================================================================
def standardize_metadata_columns(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    if 'filepath' in df.columns: df['filepath'] = df['filepath'].astype(str).str.replace('\\', '/', regex=False)
    if 'filename' not in df.columns:
        if 'filepath' in df.columns: df['filename'] = df['filepath'].apply(lambda p: os.path.basename(str(p)))
    if 'label' not in df.columns and 'filepath' in df.columns: df['label'] = df['filepath'].str.split('/').str[1]
    df['label'] = df['label'].astype(str)

    # =====> THIS IS THE CORRECTED SECTION <=====
    # Restored the original age group definitions to match the trained model checkpoint.
    if 'age_group' not in df.columns and 'age' in df.columns:
        age_num = pd.to_numeric(df['age'], errors='coerce')
        bins, names = [-1, 17, 29, 44, 64, 200], ['child', 'young_adult', 'adult', 'middle_aged', 'senior']
        df['age_group'] = pd.cut(age_num, bins=bins, labels=names).astype(str).fillna('unknown')

    if 'gender' not in df.columns and 'dominant_gender' in df.columns:
        g = df['dominant_gender'].astype(str).str.strip().str.lower()
        df['gender'] = g.map({'man':'male','male':'male','woman':'female','female':'female'}).fillna('unknown')
    if 'skin_tone' not in df.columns and 'dominant_race' in df.columns:
        df['skin_tone'] = df['dominant_race'].astype(str).str.strip().str.lower()
    return df

class MetaVocab:
    def __init__(self, df_train, cols=('age_group', 'gender', 'skin_tone')):
        self.cols, self.unk_token = cols, "<UNK>"
        self.vocabs = {c: {v: i for i, v in enumerate(sorted([self.unk_token] + df_train[c].dropna().astype(str).unique().tolist()))} for c in cols}
    def encode_row(self, row): return {c: self.vocabs[c].get(str(row[c]), self.vocabs[c][self.unk_token]) for c in self.cols}

class MetadataEncoder(nn.Module):
    def __init__(self, meta_vocab, hidden=128, out_dim=256, dropout=0.1):
        super().__init__()
        sizes = {c: len(v) for c, v in meta_vocab.vocabs.items()}
        emb_dims = {'age_group': min(16, max(4, sizes['age_group'])), 'gender': min(8, max(3, sizes['gender'])), 'skin_tone': min(16, max(4, sizes['skin_tone']))}
        self.emb_age, self.emb_gen, self.emb_tone = nn.Embedding(sizes['age_group'], emb_dims['age_group']), nn.Embedding(sizes['gender'], emb_dims['gender']), nn.Embedding(sizes['skin_tone'], emb_dims['skin_tone'])
        self.mlp = nn.Sequential(nn.Linear(sum(emb_dims.values()), hidden), nn.GELU(), nn.Dropout(dropout), nn.Linear(hidden, out_dim), nn.GELU())
    def forward(self, meta_idx): return self.mlp(torch.cat([self.emb_age(meta_idx['age_group']), self.emb_gen(meta_idx['gender']), self.emb_tone(meta_idx['skin_tone'])], dim=-1))

class DINOv3MultimodalConcat(nn.Module):
    def __init__(self, model_name, num_classes, meta_vocab, meta_hidden_dim=128, meta_out_dim=256, meta_dropout=0.3):
        super().__init__()
        model_config = AutoConfig.from_pretrained(model_name)
        model_config.output_attentions = True
        self.backbone = AutoModel.from_pretrained(model_name, config=model_config)
        image_feature_dim = self.backbone.config.hidden_size
        self.meta_encoder = MetadataEncoder(meta_vocab, hidden=meta_hidden_dim, out_dim=meta_out_dim, dropout=meta_dropout)
        self.classifier = nn.Sequential(nn.Linear(image_feature_dim + meta_out_dim, 512), nn.BatchNorm1d(512), nn.GELU(), nn.Dropout(0.5), nn.Linear(512, num_classes))
    def forward(self, pixel_values, meta_idx, output_attentions=False):
        outputs = self.backbone(pixel_values=pixel_values, output_attentions=output_attentions)
        img_features = outputs.last_hidden_state[:, 0]
        meta_features = self.meta_encoder(meta_idx)
        logits = self.classifier(torch.cat((img_features, meta_features), dim=1))
        return (logits, outputs.attentions) if output_attentions else logits

from sklearn.model_selection import StratifiedKFold
def build_kfold_splits(df_std, n_splits, seed=42):
    df, y = df_std.copy().reset_index(drop=True), df_std['label'].astype(str).values
    skf, folds = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=seed), []
    for _, test_idx in skf.split(np.arange(len(df)), y):
        dfi = df.copy(); dfi['split'] = 'Train'; dfi.loc[test_idx, 'split'] = 'Test'; folds.append(dfi)
    return folds

# =================================================================================================
# 3. ATTENTION MAP VISUALIZATION
# =================================================================================================
def get_model_and_data(config):
    df_std = standardize_metadata_columns(pd.read_csv(config.ORIGINAL_METADATA_CSV))
    fold_1_df = build_kfold_splits(df_std, config.N_SPLITS)[0]
    df_train, df_test = fold_1_df[fold_1_df['split'] == 'Train'], fold_1_df[fold_1_df['split'] == 'Test']
    classes = sorted(df_train['label'].astype(str).unique())
    class_to_idx = {c: i for i, c in enumerate(classes)}
    idx_to_class = {i: c for c, i in class_to_idx.items()}
    meta_vocab, num_classes = MetaVocab(df_train), len(classes)
    model = DINOv3MultimodalConcat(model_name=config.MODEL_NAME, num_classes=num_classes, meta_vocab=meta_vocab).to(config.DEVICE)
    model.load_state_dict(torch.load(config.LOAD_MODEL_PATH, map_location=config.DEVICE)); model.eval()
    print(f"Model and data loaded successfully.")
    return model, df_test, meta_vocab, idx_to_class

def generate_attention_map(model, image_tensor, meta_idx):
    input_tensor = image_tensor.unsqueeze(0).to(config.DEVICE)
    meta_idx_tensor = {k: torch.tensor([v], device=config.DEVICE) for k, v in meta_idx.items()}
    with torch.no_grad():
        logits, attentions = model(input_tensor, meta_idx_tensor, output_attentions=True)
    prediction_idx = logits.argmax(1).item()
    avg_attention = torch.mean(attentions[-1].squeeze(0), dim=0)
    num_patches, num_special_tokens = 196, avg_attention.shape[0] - 196
    image_patch_attention = avg_attention[0, num_special_tokens:].cpu().numpy()
    h = w = int(num_patches**0.5)
    attention_map = cv2.normalize(image_patch_attention.reshape(h, w), None, 0, 255, cv2.NORM_MINMAX)
    attention_map = cv2.resize(attention_map, (config.IMG_SIZE, config.IMG_SIZE)).astype(np.uint8)
    img_np = np.clip((input_tensor.squeeze().cpu().permute(1, 2, 0).numpy() * 0.225) + 0.45, 0, 1) # Simplified denorm
    heatmap = cv2.applyColorMap(attention_map, cv2.COLORMAP_JET)
    overlay = cv2.addWeighted((img_np * 255).astype(np.uint8), 0.6, heatmap, 0.4, 0)
    return overlay, prediction_idx

# =================================================================================================
# 4. MAIN EXECUTION
# =================================================================================================
def main():
    if not HAVE_TRANSFORMERS:
        print("Missing libraries. Please run: pip install transformers opencv-python"); return

    model, df_test, meta_vocab, idx_to_class = get_model_and_data(config)
    eval_transform = transforms.Compose([transforms.Resize((config.IMG_SIZE, config.IMG_SIZE)), transforms.ToTensor(), transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])])

    # ============================================================================================
    # =====> EDIT THIS LIST TO DEFINE YOUR COMPARISONS <=====
    # ============================================================================================
    plotting_jobs = [
        {
            "title": "Comparison by Skin Tone (Male, Adult)",
            "save_filename": "attention_comparison_skin_tone.png",
            "plot_grid": (2, 3), # Rows, Columns
            "image_criteria": [
                {"gender": "male", "age_group": "adult", "skin_tone": "white"},
                {"gender": "male", "age_group": "adult", "skin_tone": "black"},
                {"gender": "male", "age_group": "adult", "skin_tone": "asian"},
            ]
        },
        {
            "title": "Comparison by Age (Female, White)",
            "save_filename": "attention_comparison_age.png",
            "plot_grid": (2, 3),
            "image_criteria": [
                {"gender": "female", "age_group": "child", "skin_tone": "white"},
                {"gender": "female", "age_group": "adult", "skin_tone": "white"},
                {"gender": "female", "age_group": "senior", "skin_tone": "white"},
            ]
        },
        {
            "title": "Comparison by Gender (Adult, Black)",
            "save_filename": "attention_comparison_gender.png",
            "plot_grid": (2, 2),
            "image_criteria": [
                {"gender": "male", "age_group": "adult", "skin_tone": "black"},
                {"gender": "female", "age_group": "adult", "skin_tone": "black"},
            ]
        },
    ]

    # --- Plotting Loop ---
    for job in plotting_jobs:
        print(f"\n--- Generating plot: {job['title']} ---")
        num_images = len(job['image_criteria'])
        rows, cols = job['plot_grid']
        fig, axes = plt.subplots(rows, cols, figsize=(5 * cols, 5 * rows))
        fig.suptitle(job['title'], fontsize=18, fontweight='bold')
        axes = axes.flatten()

        for i, criteria in enumerate(job['image_criteria']):
            # Find a matching image
            subset = df_test
            for key, value in criteria.items():
                subset = subset[subset[key] == value]

            if subset.empty:
                print(f"Warning: No image found for criteria: {criteria}. Skipping.")
                axes[i].text(0.5, 0.5, f"Image not found\n{criteria}", ha='center', va='center')
                axes[i].axis('off')
                axes[i + cols].axis('off')
                continue

            sample = subset.sample(1).iloc[0]

            # Generate attention map
            img_path = os.path.join(config.DATA_DIR, sample['filepath'])
            img_pil = Image.open(img_path).convert("RGB")
            img_tensor, meta_idx = eval_transform(img_pil), meta_vocab.encode_row(sample)
            overlay, pred_idx = generate_attention_map(model, img_tensor, meta_idx)

            # Get info for titles
            pred_label, true_label = idx_to_class[pred_idx], sample['label']
            subtitle = f"True: {true_label}, Pred: {pred_label}\n" + ", ".join([f"{k.title()}: {v}" for k, v in criteria.items()])

            # Plot original and overlay
            axes[i].imshow(img_pil)
            axes[i].set_title(subtitle)
            axes[i].axis('off')

            axes[i + cols].imshow(overlay)
            axes[i + cols].axis('off')

        # Remove any unused subplots
        for i in range(num_images, cols):
            axes[i].axis('off')
            axes[i + cols].axis('off')

        plt.tight_layout(rect=[0, 0, 1, 0.96])
        plt.savefig(job['save_filename'])
        plt.show()
        print(f"Plot saved as {job['save_filename']}")

if __name__ == '__main__':
    main()
```

```python
# =================================================================================================
# Attention Map Visualization Script for Multiple Subgroup Comparisons
# - Generates separate plots for different demographic axes (age, skin tone, etc.)
# - Easily configurable via the `plotting_jobs` list in the main section.
# =================================================================================================

import os
import random
import warnings
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
from PIL import Image
from torchvision import transforms
import cv2

# Conditional import for Hugging Face transformers
try:
    from transformers import AutoModel, AutoConfig
    HAVE_TRANSFORMERS = True
except ImportError:
    HAVE_TRANSFORMERS = False

# Suppress warnings for cleaner output
warnings.filterwarnings('ignore')
torch.backends.cudnn.benchmark = True


# =================================================================================================
# 1. CONFIGURATION
# =================================================================================================
class Config:
    DATA_DIR = r"D:\AIUB\Neural\Synthetic_Skin_Disease_Dataset"
    ORIGINAL_METADATA_CSV = r"D:\AIUB\Neural\complete_metadata_m.csv"
    RESULTS_DIR = "new_results_Multimodal_DINO_Concat_KFold"
    LOAD_MODEL_PATH = os.path.join(RESULTS_DIR, "best_model_fold_5.pth")
    MODEL_NAME = "facebook/dinov3-vitb16-pretrain-lvd1689m"
    IMG_SIZE = 224
    DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
    N_SPLITS = 7

config = Config()
print(f"Using device: {config.DEVICE}")


# =================================================================================================
# 2. MODEL AND DATA HELPERS (from previous versions)
# =================================================================================================
def standardize_metadata_columns(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    if 'filepath' in df.columns: df['filepath'] = df['filepath'].astype(str).str.replace('\\', '/', regex=False)
    if 'filename' not in df.columns:
        if 'filepath' in df.columns: df['filename'] = df['filepath'].apply(lambda p: os.path.basename(str(p)))
    if 'label' not in df.columns and 'filepath' in df.columns: df['label'] = df['filepath'].str.split('/').str[1]
    df['label'] = df['label'].astype(str)

    # =====> THIS IS THE CORRECTED SECTION <=====
    # Restored the original age group definitions to match the trained model checkpoint.
    if 'age_group' not in df.columns and 'age' in df.columns:
        age_num = pd.to_numeric(df['age'], errors='coerce')
        bins, names = [-1, 17, 29, 44, 64, 200], ['child', 'young_adult', 'adult', 'middle_aged', 'senior']
        df['age_group'] = pd.cut(age_num, bins=bins, labels=names).astype(str).fillna('unknown')

    if 'gender' not in df.columns and 'dominant_gender' in df.columns:
        g = df['dominant_gender'].astype(str).str.strip().str.lower()
        df['gender'] = g.map({'man':'male','male':'male','woman':'female','female':'female'}).fillna('unknown')
    if 'skin_tone' not in df.columns and 'dominant_race' in df.columns:
        df['skin_tone'] = df['dominant_race'].astype(str).str.strip().str.lower()
    return df

class MetaVocab:
    def __init__(self, df_train, cols=('age_group', 'gender', 'skin_tone')):
        self.cols, self.unk_token = cols, "<UNK>"
        self.vocabs = {c: {v: i for i, v in enumerate(sorted([self.unk_token] + df_train[c].dropna().astype(str).unique().tolist()))} for c in cols}
    def encode_row(self, row): return {c: self.vocabs[c].get(str(row[c]), self.vocabs[c][self.unk_token]) for c in self.cols}

class MetadataEncoder(nn.Module):
    def __init__(self, meta_vocab, hidden=128, out_dim=256, dropout=0.1):
        super().__init__()
        sizes = {c: len(v) for c, v in meta_vocab.vocabs.items()}
        emb_dims = {'age_group': min(16, max(4, sizes['age_group'])), 'gender': min(8, max(3, sizes['gender'])), 'skin_tone': min(16, max(4, sizes['skin_tone']))}
        self.emb_age, self.emb_gen, self.emb_tone = nn.Embedding(sizes['age_group'], emb_dims['age_group']), nn.Embedding(sizes['gender'], emb_dims['gender']), nn.Embedding(sizes['skin_tone'], emb_dims['skin_tone'])
        self.mlp = nn.Sequential(nn.Linear(sum(emb_dims.values()), hidden), nn.GELU(), nn.Dropout(dropout), nn.Linear(hidden, out_dim), nn.GELU())
    def forward(self, meta_idx): return self.mlp(torch.cat([self.emb_age(meta_idx['age_group']), self.emb_gen(meta_idx['gender']), self.emb_tone(meta_idx['skin_tone'])], dim=-1))

class DINOv3MultimodalConcat(nn.Module):
    def __init__(self, model_name, num_classes, meta_vocab, meta_hidden_dim=128, meta_out_dim=256, meta_dropout=0.3):
        super().__init__()
        model_config = AutoConfig.from_pretrained(model_name)
        model_config.output_attentions = True
        self.backbone = AutoModel.from_pretrained(model_name, config=model_config)
        image_feature_dim = self.backbone.config.hidden_size
        self.meta_encoder = MetadataEncoder(meta_vocab, hidden=meta_hidden_dim, out_dim=meta_out_dim, dropout=meta_dropout)
        self.classifier = nn.Sequential(nn.Linear(image_feature_dim + meta_out_dim, 512), nn.BatchNorm1d(512), nn.GELU(), nn.Dropout(0.5), nn.Linear(512, num_classes))
    def forward(self, pixel_values, meta_idx, output_attentions=False):
        outputs = self.backbone(pixel_values=pixel_values, output_attentions=output_attentions)
        img_features = outputs.last_hidden_state[:, 0]
        meta_features = self.meta_encoder(meta_idx)
        logits = self.classifier(torch.cat((img_features, meta_features), dim=1))
        return (logits, outputs.attentions) if output_attentions else logits

from sklearn.model_selection import StratifiedKFold
def build_kfold_splits(df_std, n_splits, seed=42):
    df, y = df_std.copy().reset_index(drop=True), df_std['label'].astype(str).values
    skf, folds = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=seed), []
    for _, test_idx in skf.split(np.arange(len(df)), y):
        dfi = df.copy(); dfi['split'] = 'Train'; dfi.loc[test_idx, 'split'] = 'Test'; folds.append(dfi)
    return folds

# =================================================================================================
# 3. ATTENTION MAP VISUALIZATION
# =================================================================================================
def get_model_and_data(config):
    df_std = standardize_metadata_columns(pd.read_csv(config.ORIGINAL_METADATA_CSV))
    fold_1_df = build_kfold_splits(df_std, config.N_SPLITS)[0]
    df_train, df_test = fold_1_df[fold_1_df['split'] == 'Train'], fold_1_df[fold_1_df['split'] == 'Test']
    classes = sorted(df_train['label'].astype(str).unique())
    class_to_idx = {c: i for i, c in enumerate(classes)}
    idx_to_class = {i: c for c, i in class_to_idx.items()}
    meta_vocab, num_classes = MetaVocab(df_train), len(classes)
    model = DINOv3MultimodalConcat(model_name=config.MODEL_NAME, num_classes=num_classes, meta_vocab=meta_vocab).to(config.DEVICE)
    model.load_state_dict(torch.load(config.LOAD_MODEL_PATH, map_location=config.DEVICE)); model.eval()
    print(f"Model and data loaded successfully.")
    return model, df_test, meta_vocab, idx_to_class

def generate_attention_map(model, image_tensor, meta_idx):
    input_tensor = image_tensor.unsqueeze(0).to(config.DEVICE)
    meta_idx_tensor = {k: torch.tensor([v], device=config.DEVICE) for k, v in meta_idx.items()}
    with torch.no_grad():
        logits, attentions = model(input_tensor, meta_idx_tensor, output_attentions=True)
    prediction_idx = logits.argmax(1).item()
    avg_attention = torch.mean(attentions[-1].squeeze(0), dim=0)
    num_patches, num_special_tokens = 196, avg_attention.shape[0] - 196
    image_patch_attention = avg_attention[0, num_special_tokens:].cpu().numpy()
    h = w = int(num_patches**0.5)
    attention_map = cv2.normalize(image_patch_attention.reshape(h, w), None, 0, 255, cv2.NORM_MINMAX)
    attention_map = cv2.resize(attention_map, (config.IMG_SIZE, config.IMG_SIZE)).astype(np.uint8)
    img_np = np.clip((input_tensor.squeeze().cpu().permute(1, 2, 0).numpy() * 0.225) + 0.45, 0, 1) # Simplified denorm
    heatmap = cv2.applyColorMap(attention_map, cv2.COLORMAP_JET)
    overlay = cv2.addWeighted((img_np * 255).astype(np.uint8), 0.6, heatmap, 0.4, 0)
    return overlay, prediction_idx

# =================================================================================================
# 4. MAIN EXECUTION
# =================================================================================================
def main():
    if not HAVE_TRANSFORMERS:
        print("Missing libraries. Please run: pip install transformers opencv-python"); return

    model, df_test, meta_vocab, idx_to_class = get_model_and_data(config)
    eval_transform = transforms.Compose([transforms.Resize((config.IMG_SIZE, config.IMG_SIZE)), transforms.ToTensor(), transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])])

    # ============================================================================================
    # =====> EDIT THIS LIST TO DEFINE YOUR COMPARISONS <=====
    # ============================================================================================
    plotting_jobs = [
        {
            "title": "Comparison by Gender (Adult, White)",
            "save_filename": "attention_comparison_gender.png",
            "plot_grid": (2, 2), # (Rows, Columns) -> 2 rows (original, heatmap), 2 columns (male, female)
            "image_criteria": [
                {"gender": "male", "age_group": "adult", "skin_tone": "white"},
                {"gender": "female", "age_group": "adult", "skin_tone": "white"},
            ]
        },
        {
            "title": "Comparison by Age Group (Male, White)",
            "save_filename": "attention_comparison_age.png",
            "plot_grid": (2, 2),
            "image_criteria": [
                {"age_group": "child", "gender": "male", "skin_tone": "white"},
                {"age_group": "adult", "gender": "male", "skin_tone": "white"},
            ]
        },
        {
            "title": "Comparison: Black vs. White Skin (Male, Adult)",
            "save_filename": "attention_comparison_black_vs_white.png",
            "plot_grid": (2, 2),
            "image_criteria": [
                {"skin_tone": "black", "gender": "male", "age_group": "adult"},
                {"skin_tone": "white", "gender": "male", "age_group": "adult"},
            ]
        },
        {
            "title": "Comparison: White vs. Asian Skin (Male, Adult)",
            "save_filename": "attention_comparison_white_vs_asian.png",
            "plot_grid": (2, 2),
            "image_criteria": [
                {"skin_tone": "white", "gender": "male", "age_group": "adult"},
                {"skin_tone": "asian", "gender": "male", "age_group": "adult"},
            ]
        },
        {
            "title": "Comparison: White vs. Indian Skin (Male, Adult)",
            "save_filename": "attention_comparison_white_vs_indian.png",
            "plot_grid": (2, 2),
            "image_criteria": [
                {"skin_tone": "white", "gender": "male", "age_group": "adult"},
                {"skin_tone": "indian", "gender": "male", "age_group": "adult"},
            ]
        },
    ]

    # --- Plotting Loop ---
    for job in plotting_jobs:
        print(f"\n--- Generating plot: {job['title']} ---")
        num_images = len(job['image_criteria'])
        rows, cols = job['plot_grid']
        fig, axes = plt.subplots(rows, cols, figsize=(5 * cols, 5 * rows))
        fig.suptitle(job['title'], fontsize=18, fontweight='bold')
        axes = axes.flatten()

        for i, criteria in enumerate(job['image_criteria']):
            # Find a matching image
            subset = df_test
            for key, value in criteria.items():
                subset = subset[subset[key] == value]

            if subset.empty:
                print(f"Warning: No image found for criteria: {criteria}. Skipping.")
                axes[i].text(0.5, 0.5, f"Image not found\n{criteria}", ha='center', va='center')
                axes[i].axis('off')
                axes[i + cols].axis('off') # Also turn off the corresponding heatmap plot
                continue

            sample = subset.sample(1).iloc[0]

            # Generate attention map
            img_path = os.path.join(config.DATA_DIR, sample['filepath'])
            img_pil = Image.open(img_path).convert("RGB")
            img_tensor, meta_idx = eval_transform(img_pil), meta_vocab.encode_row(sample)
            overlay, pred_idx = generate_attention_map(model, img_tensor, meta_idx)

            # Get info for titles
            pred_label, true_label = idx_to_class[pred_idx], sample['label']
            subtitle = f"True: {true_label}, Pred: {pred_label}\n" + ", ".join([f"{k.replace('_', ' ').title()}: {v}" for k, v in criteria.items()])

            # Plot original and overlay
            # The script plots originals on the first row and heatmaps on the second.
            axes[i].imshow(img_pil)
            axes[i].set_title(subtitle)
            axes[i].axis('off')

            axes[i + cols].imshow(overlay)
            axes[i + cols].set_title(f"Attention Map for Image {i+1}")
            axes[i + cols].axis('off')

        # Remove any unused subplots
        for i in range(num_images, cols):
            axes[i].axis('off')
            axes[i + cols].axis('off')

        plt.tight_layout(rect=[0, 0, 1, 0.96])
        plt.savefig(job['save_filename'])
        plt.show()
        print(f"Plot saved as {job['save_filename']}")

if __name__ == '__main__':
    main()
```
